```python
import pandas as pd
df = pd.read_csv("wound_data.csv")
df.head()
```

```python
df.shape # rows and columns
df.columns # column names
df.info() # dtypes + missing values
df.describe() # simple statistics
```

```python
import numpy as np #numpy is a library for numerical computing
readings = np.array([36.5, 37.2, 38.1]) #accepts an array of numbers

print("Tempreture Mean:", readings.mean())
print(readings > 37.7)

temp = df["temperature"].to_numpy() #tonumpy() converts a pandas series to a numpy array
ph = df["pH"].to_numpy()

correlation = np.corrcoef(temp, ph)[0, 1] #np.corrcoef() calculates the correlation coefficient between two arrays
print("Correlation:", round(correlation, 2))
```

```python
df["high_temperature"] = df["temperature"] > 37.7 #df[] is used to create a new column in the dataframe
df["high_pH"] = df["pH"] > 7.65
df["combined_risk"] = ( #df[] can even combine multiple columns to create a new column, very nice
 0.65 * (df["temperature"] - 36.5)
 + 0.35 * (df["pH"] - 7.0) * 2
)
# these are examples of data manipulation, which is important for machine learning as AI handles data better this way. 
```

```python
# Classification: a two-condition prediction rule
df["prediction"] = np.where( #df[] made a new column, np.where() is a conditional function that returns one value if the condition is true/false
 (df["temperature"] > 37.7) & (df["pH"] > 7.65),
 "deteriorating", "stable"
)

df["correct"] = df["prediction"] == df["status"]
accuracy = df["correct"].mean()
print("Accuracy:", round(accuracy, 2)) 
# Save the updated table as a new CSV file
df.to_csv("wound_data_with_predictions.csv",
index=False)

mistakes = df[df["correct"] == False]
mistakes[["patient_id", "temperature",
 "pH", "status", "prediction"]]
```

```python
# Open the CSV that was already saved
df = pd.read_csv("wound_data_with_predictions.csv")

# Create a new column
new_column = pd.DataFrame({
 "risk_level": np.where( (df["temperature"] > 37.7) & (df["pH"] > 7.65),
 "high", "low"
 )
})

# Append the new column to the DataFrame
df = pd.concat([df, new_column], axis=1) #axis means stack the new column horizontally (axis=1) instead of vertically (axis=0)

# Drop a column
df = df.drop(columns=["correct"]) #df.drop() removes a column from the DataFrame

# Remove one value by replacing it with a missing value
df.loc[df["patient_id"] == 9, "pH"] = np.nan #this line replaces the pH value for patient_id 9 with NaN (missing value)

# Save the final updated version
df.to_csv("wound_data_updated.csv", index=False)
print(df)
```

```python
from sklearn.cluster import KMeans
X = df[["temperature", "pH"]]
kmeans = KMeans(n_clusters=3, random_state=7) #cluster = 3 means to group the data into 3 clusters. 
df["cluster"] = kmeans.fit_predict(X)
print(df[["patient_id", "temperature", "pH",
"cluster"]])
```

```python
df = pd.read_csv("wound_data_with_predictions.csv")
df_rulefail = df.copy()

# Deliberately create one prediction that does not follow the rule
df_rulefail.loc[
 df_rulefail["patient_id"] == 10,
 "prediction"
] = "deteriorating"

# Calculate what the prediction should be according to the rule
df_rulefail["rule_prediction"] = np.where(
 (df_rulefail["temperature"] > 37.7)
 & (df_rulefail["pH"] > 7.65),
 "deteriorating",
 "stable"
)

# Compare the stored prediction with the rule-based prediction
df_rulefail["follows_rule"] = (
 df_rulefail["prediction"]
 == df_rulefail["rule_prediction"]
)

# Display only rule failures
df_rulefail[df_rulefail["follows_rule"] == False]
#this chunk of code is used if a rule exists which the algorithm follows. 
```
